```python
import numpy as np
import matplotlib.pyplot as plt

# Data kept as a dictionary to preserve labels/keys
points = {
    0: [0.1, 0.9],    # 0
    1: [0.3, 0.7],    # 1
    2: [0.35, 0.65],  # 2
    3: [0.4, 0.6],    # 3
    4: [0.42, 0.58],  # 4
    5: [0.45, 0.55],  # 5
    6: [0.6, 0.4],    # 6
    7: [0.8, 0.2],    # 7
    8: [0.96, 0.04],  # 8
    9: [0.99, 0.01]   # 9
}

reference = np.array([0.0, 0.0])

colormap = {
    0: 'blue',
    1: 'red',
    2: 'green',
    3: 'orange',
    4: 'purple',
    5: 'brown',
    6: 'pink',
    7: 'gray',
    8: 'olive',
    9: 'cyan'
}

def hypervolume_2d(pts_dict, ref):
    """
    Hypervolume dla 2D, maksymalizacja.
    Sortuje punkty po x, aby zapewnić poprawne pole pod "schodkami".
    """
    sorted_pts = sorted(pts_dict.values(), key=lambda p: p[0])

    hv = 0.0
    previous_x = ref[0]

    for x, y in sorted_pts:
        hv += (x - previous_x) * (y - ref[1])
        previous_x = x

    return hv

def hv_contribution(label, current_points, ref):
    """
    Wyłączny wkład punktu do HV dla aktualnego zbioru punktów.
    """
    hv_all = hypervolume_2d(current_points, ref)
    # Create a new dictionary excluding the point with the current label
    points_without = {k: v for k, v in current_points.items() if k != label}
    hv_without = hypervolume_2d(points_without, ref)

    return hv_all - hv_without

def plot_on_ax(ax, iteration, current_points, contrib, ref):
    """
    Rysuje wykres na przekazanym obiekcie Axes.
    """
    # Rysowanie dominowanych prostokątów
    for label, (x, y) in current_points.items():
        ax.fill(
            [0, x, x, 0],
            [0, 0, y, y],
            alpha=0.85,
            color=colormap[label],
            edgecolor='gray'
        )

    # Punkty
    if current_points:
        x_coords = [p[0] for p in current_points.values()]
        y_coords = [p[1] for p in current_points.values()]
        ax.scatter(
            x_coords,
            y_coords,
            s=20,
            color='black',
            zorder=3
        )

    # Opisy punktów
    for label, (x, y) in current_points.items():
        ax.text(
            x + 0.015,
            y + 0.015,
            f"{label}: HV={contrib[label]:.4f}",
            fontsize=7
        )

    # Reference point
    ax.scatter(
        [ref[0]],
        [ref[1]],
        marker="x",
        s=100,
        c='red',
        label="Ref point (0,0)",
        zorder=4
    )

    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_title(f"Iteration {iteration}")

    ax.set_xlim(-0.05, 1.2)
    ax.set_ylim(-0.05, 1.2)

    ax.grid(True, linestyle='--', alpha=0.6)
    if iteration == 0:
        ax.legend(loc='upper right', fontsize=8)


# --------------------------
# Główna pętla iteracyjna
# --------------------------

# Tworzymy siatkę wykresów: 2 wiersze, 3 kolumny dla 6 iteracji
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten() # Spłaszcza tablicę 2D do 1D dla łatwiejszego iterowania

for iteration in range(6): # Runs for 0, 1, 2, 3, 4, 5
    
    # Oblicz wkłady na podstawie aktualnych punktów
    contrib = {
        label: hv_contribution(label, points, reference)
        for label in points.keys()
    }
    
    print(f"\n--- Iteration {iteration} ---")
    for label, c in contrib.items():
        print(f"Point {label}: {points[label]} -> {c:.8f}")

    # Rysuj aktualny stan na odpowiednim subplocie
    plot_on_ax(axes[iteration], iteration, points, contrib, reference)
    
    # Usuń punkt o najmniejszym wkładzie (oprócz ostatniej iteracji)
    if iteration < 5:
        # Znajdź klucz z najmniejszą wartością w słowniku contrib
        min_label = min(contrib, key=contrib.get)
        print(f">>> Removing Point {min_label} (lowest contribution: {contrib[min_label]:.8f})")
        
        # Usuń punkt ze słownika głównego
        del points[min_label]

plt.tight_layout() # Automatycznie dostosowuje odstępy między wykresami
plt.show()
```
